## Step 1: Introduction

```python
Can i predict outage severity based on previous given outage events. 

    The regression angle — "Predicting outage severity (duration and customers affected) given an outage event" — is more impactful for these reasons:
For employers: Predicting how bad something will be is directly actionable. A utility company, grid operator, or government agency can use severity predictions to pre-position repair crews, allocate resources, and prioritize response. Cause classification is more academic by comparison.
For your portfolio: It lets you tell a clean story — "I built a model that predicts how many customers will be affected and for how long, which could reduce response time and economic loss." That narrative resonates with both technical and non-technical hiring managers.
Technically it's richer too:

You deal with skewed targets (most outages are short, a few are catastrophic) → log transformation, robust evaluation
You can frame it as two tasks: predict duration AND customers affected, then see if they share the same important features
Natural extension into causal inference — what factors cause longer outages vs just correlate? That's a very hot skill right now
Could add a clustering component to find outage archetypes (e.g. "short high-impact" vs "long low-impact")

The causal inference angle in particular is something most DS candidates don't touch and would make you stand out significantly.
```

## Step 2: Data Cleaning and Exploratory Data Analysis

```python
cause_counts = df['CAUSE.CATEGORY'].value_counts().reset_index()
cause_counts.columns = ['Cause', 'Count']

fig = cause_counts.plot(x='Cause', y='Count', kind='bar',
                        title='Power Outages by Cause Category')
#fig.write_image("../docs/plots/outage_cause_counts.png")
```
